```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import shap
from google.colab import drive
from sklearn.ensemble import RandomForestClassifier

# 1. Mount Google Drive
drive.mount('/content/drive')

# 2. Load Dataset
dataset_path = '/content/drive/MyDrive/Colab/windowed.csv'
df = pd.read_csv(dataset_path)

if 'class' in df.columns and 'user' in df.columns:
    X = df.drop(columns=['class', 'user'])
    y = df['class']
    groups = df['user']
    unique_users = df['user'].unique()
    class_names = [str(c) for c in np.sort(np.unique(y))]
    print(f"Dataset loaded. Classes: {class_names}")
else:
    raise ValueError("Columns 'class' or 'user' not found in dataset.")

# 3. Train/Test Split (Leave-One-User-Out approach)
# We train on all users except one to simulate an unseen subject
test_user = unique_users[0]
train_idx = df['user'] != test_user
test_idx = df['user'] == test_user

X_train, y_train = X[train_idx], y[train_idx]
X_test, y_test = X[test_idx], y[test_idx]

print(f"Training on all users except: {test_user}")
model = RandomForestClassifier(n_estimators=150, max_depth=20, random_state=42, n_jobs=-1)
model.fit(X_train, y_train)

# 4. Get Predictions
y_pred = model.predict(X_test)
y_test_reset = y_test.reset_index(drop=True)
X_test_reset = X_test.reset_index(drop=True)

# 5. SHAP Calculation
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_test_reset)

# Note: In multi-class, shap_values is a list of arrays (one per class)
# or a 3D array [samples, features, classes] depending on version.
is_list = isinstance(shap_values, list)

# 6. Local Explanation Loop
for class_idx, class_label in enumerate(np.sort(y.unique())):
    print(f"\n" + "="*50)
    print(f" ANALYZING CLASS: {class_label}")
    print("="*50)

    # Indices for samples belonging to this class
    class_mask = (y_test_reset == class_label)

    # 1. Find a Correct Prediction
    correct_indices = np.where(class_mask & (y_pred == class_label))[0]
    # 2. Find an Incorrect Prediction
    incorrect_indices = np.where(class_mask & (y_pred != class_label))[0]

    scenarios = [
        ("Correct Prediction", correct_indices),
        ("Incorrect Prediction", incorrect_indices)
    ]

    for title, indices in scenarios:
        if len(indices) > 0:
            idx = indices[0] # Take the first available sample
            actual_pred = y_pred[idx]

            print(f"\n>>> {title} (Target: {class_label}, Predicted: {actual_pred})")
            print("Feature Values for this sample:")
            print(X_test_reset.iloc[idx].to_frame().T)

            # Prepare SHAP values for the Waterfall Plot
            # We explain the prediction relative to the target class
            if is_list:
                current_shap = shap_values[class_idx][idx]
                expected_val = explainer.expected_value[class_idx]
            else:
                current_shap = shap_values[idx, :, class_idx]
                expected_val = explainer.expected_value[class_idx]

            # Create explanation object
            exp = shap.Explanation(
                values=current_shap,
                base_values=expected_val,
                data=X_test_reset.iloc[idx].values,
                feature_names=X.columns.tolist()
            )

            # Plot
            plt.figure(figsize=(10, 6))
            shap.plots.waterfall(exp, show=False)
            plt.title(f"{title}: Class {class_label} (Sample Index {idx})")
            plt.show()
        else:
            print(f"\n>>> No samples found for '{title}' in Class {class_label}")
```
